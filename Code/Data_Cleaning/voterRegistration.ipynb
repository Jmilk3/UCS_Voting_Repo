# Import statements

```python
import pandas as pd
import numpy as np
```

# Creating subsets

```python
voterCSV = pd.read_csv('/content/NC_Voter_Reg_Nov_23.csv')
voterCSV
```

# Filter for Johnston County (Smithfield)

```python
countyJohnston = voterCSV[voterCSV['county_desc'] == 'JOHNSTON']
countyJohnston = countyJohnston.drop('vtd_abbrv', axis = 1)
countyJohnston
```

# Filter for Mecklenberg County (Charlotte)

```python
countyMecklenberg = voterCSV[voterCSV['county_desc'] == 'MECKLENBURG']
countyMecklenberg['precinct_abbrv'] = countyMecklenberg['precinct_abbrv'].str.replace('PR','', regex = False)
countyMecklenberg = countyMecklenberg.drop('vtd_abbrv', axis = 1)
countyMecklenberg
```

```python
'''
We read the csv
'''
voterCSV2022 = pd.read_csv('/content/Registration_Data_July-2022.csv')
countyMecklenberg2022 = voterCSV2022[voterCSV2022['county_desc'] == 'MECKLENBURG']
countyMecklenberg2022 = countyMecklenberg2022.drop('vtd_abbrv', axis = 1)
countyMecklenberg2022
```

# Filter for Randolph County (Asheboro)

```python
countyRandolph = voterCSV[voterCSV['county_desc'] == 'RANDOLPH']
countyRandolph['precinct_abbrv'] = countyRandolph['precinct_abbrv'].str.replace('PR','', regex = False)
countyRandolph = countyRandolph.drop('vtd_abbrv', axis = 1)
countyRandolph
```

# We download the csvs

```python
'''
This is our download command for loading the filtered csv once we have decided which csvs we want to keep.
'''
countyJohnston = countyJohnston.to_csv('Smithfield_CSV.csv')
countyMecklenberg = countyMecklenberg.to_csv('Charlotte_CSV.csv')
countyRandolph = countyRandolph.to_csv('Asheboro_CSV.csv')
countyMecklenberg2022 = countyMecklenberg2022.to_csv('Charlotte_2022_CSV.csv')
```
